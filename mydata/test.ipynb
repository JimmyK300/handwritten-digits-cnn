```python
import torch
import matplotlib.pyplot as plt
import math as m
```

```python
file = "testImage\Minh Ngo\minh_ngo_custom_mnist.bin"
with open(file, "rb") as f:
    data = f.read()
    # print(data[2])
    # print("ok")
```

```python
size = int(len(data)/(28*28))
classes = 10
size_per_class = int(size/classes)
```

```python
images = torch.tensor([[(255-x)/255. for x in data[i*28*28:(i+1)*28*28]] for i in range(size)]).reshape(-1,28,28)
labels = [i//size_per_class for i in range(size)]
images.shape
```

```python
idx = torch.randint(0, size, (1,)).item()

image = images[idx]
label = labels[idx]

print(f"Label = {label}")
plt.imshow(image, cmap="grey")
plt.axis("off")
plt.show()
```
